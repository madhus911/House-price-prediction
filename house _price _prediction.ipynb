```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.histplot(df['SalePrice'],kde=True)
plt.title('Distribution of House prices')
plt.xlabel('Sale price')
plt.ylabel('Frequency')
plt.show()
```

```python
none_cols=['PoolQC','MiscFeature','Alley','Fence','FireplaceQu','GarageType','GarageFinish','GarageQual','GarageCond','BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1','BsmtFinType2','MasVnrType']
df[none_cols]=df[none_cols].fillna('None')
```

```python
zero_cols=['GarageYrBlt','GarageCars','GarageArea','BsmtFinSF1','BsmtFinSF2','BsmtUnfSF','TotalBsmtSF','BsmtFullBath','BsmtHalfBath','MasVnrArea']
df[zero_cols]=df[zero_cols].fillna(0)
```

```python
df['LotFrontage']=df['LotFrontage'].fillna(df['LotFrontage'].median())
```

```python
missing=df.isnull().sum()
missing=missing[missing>0].sort_values(ascending=False)
missing
```

```python
df['Electrical']=df['Electrical'].fillna(df['Electrical'].mode()[0])
```

```python
df.isnull().sum().sum()
```

```python
X=df.drop('SalePrice',axis=1)
y=df['SalePrice']
print(X.shape)
print(y.shape)
```

```python
X=X.drop('Id',axis=1)
```

```python
X.head()
```

```python
X.dtypes
```

```python
Categorical_cols=X.select_dtypes(include=['str']).columns
print(len(Categorical_cols))
print(Categorical_cols.tolist())
```

```python
X=pd.get_dummies(X,drop_first=True)
print(X.shape)
```

```python
print('Missing values:',X.isnull().sum().sum())
```

```python
correlation=df.corr(numeric_only=True)['SalePrice'].sort_values(ascending=False)
correlation.head(15)
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x='GrLivArea',y='SalePrice')
plt.title('Living area vs House price')
plt.xlabel('Above Ground living area')
plt.ylabel('Sale price')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
comparision= pd.DataFrame({'Actual Price':y_test.values,'Predicted Price':y_pred})
comparision.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)
print('MAE:',mae)
print('MSE:',mse)
print('RMSE:',rmse)
print('R2 SCORE:',r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_model=RandomForestRegressor(n_estimators=200,random_state=42,n_jobs=-1)
rf_model.fit(X_train,y_train)
```

```python
rf_pred=rf_model.predict(X_test)
```

```python
rf_mae=mean_absolute_error(y_test,rf_pred)
rf_mse=mean_squared_error(y_test,rf_pred)
rf_rmse=np.sqrt(rf_mse)
rf_r2=r2_score(y_test,rf_pred)
print('MAE:',rf_mae)
print('MSE:',rf_mse)
print('RMSE:',rf_rmse)
print('R2 SCORE:',rf_r2)
```

```python
feature_importance=pd.DataFrame({'Feature':X_train.columns,'Importance':rf_model.feature_importances_})
feature_importance=feature_importance.sort_values(by='Importance',ascending=False)
feature_importance.head(15)
```

```python
plt.figure(figsize=(10,7))
top_features=feature_importance.head(15)
sns.barplot(data=top_features,x='Importance',y='Feature')
plt.title('Top 15 Important features')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
```

```python
results=pd.DataFrame({'Model':['Linear Regression','Random Forest'],'MAE':[mae,rf_mae],'MSE':[mse,rf_mse],'RMSE':[rmse,rf_rmse],'R2_score':[r2,rf_r2]})
results
```

```python
plt.figure(figsize=(10,7))
sns.barplot(data=results,x='Model',y='R2_score')
plt.title('Model R2 score comaprision')
plt.xlabel('R2 score')
plt.ylabel('Model')
plt.ylim(0,1)
plt.show()
```
